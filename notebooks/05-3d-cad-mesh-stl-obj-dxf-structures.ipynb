## 1. Binary STL Binary Layout Specification

A binary STL file comprises an 80-byte header, followed by a 32-bit unsigned integer triangle count, and consecutive 50-byte facet records:

| Field Name | Data Type | Size (Bytes) | Description |
|:---|:---:|:---:|:---|
| **Header** | `char[80]` | 80 | ASCII metadata (cannot start with `solid`) |
| **Number of Triangles** | `uint32` | 4 | Little-endian facet count $N$ |
| **Facet Normal Vector** | `float32[3]` | 12 | Normal $\vec{n} = (n_x, n_y, n_z)$ |
| **Vertex 1 Coordinates** | `float32[3]` | 12 | Cartesian $(v_{1x}, v_{1y}, v_{1z})$ |
| **Vertex 2 Coordinates** | `float32[3]` | 12 | Cartesian $(v_{2x}, v_{2y}, v_{2z})$ |
| **Vertex 3 Coordinates** | `float32[3]` | 12 | Cartesian $(v_{3x}, v_{3y}, v_{3z})$ |
| **Attribute Byte Count** | `uint16` | 2 | Typically `0x0000` (or color flags) |

Total binary file size is precisely:

$$\text{FileSize} = 84 + (50 \times N) \text{ bytes}$$


```python
import struct

def parse_synthetic_stl_facet():
    # Construct synthetic 50-byte binary facet record (Normal + 3 Vertices + 2 byte attr)
    normal = (0.0, 0.0, 1.0)
    v1 = (0.0, 0.0, 0.0)
    v2 = (10.0, 0.0, 0.0)
    v3 = (0.0, 10.0, 0.0)
    attr = 0

    facet_bytes = struct.pack("<3f 3f 3f 3f H", *normal, *v1, *v2, *v3, attr)
    print(f"Generated Binary STL Facet Record: {len(facet_bytes)} bytes")
    
    # Calculate triangle area via cross product
    # vA = v2 - v1 = (10, 0, 0), vB = v3 - v1 = (0, 10, 0)
    # Area = 0.5 * |vA x vB| = 0.5 * (10 * 10) = 50.0 sq units
    area = 0.5 * (10.0 * 10.0)
    print(f"Facet Normal: {normal}")
    print(f"Calculated 3D Surface Area: {area:.2f} sq mm")

parse_synthetic_stl_facet()
```
