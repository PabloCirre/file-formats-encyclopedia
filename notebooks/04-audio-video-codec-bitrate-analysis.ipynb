## 1. Nyquist-Shannon Theorem & Frequency Cutoff Analysis

The Nyquist-Shannon sampling theorem states that perfect acoustic reconstruction requires a sampling frequency $f_s > 2 f_{\text{max}}$:

$$f_{\text{Nyquist}} = \frac{f_s}{2}$$

| Sampling Rate ($f_s$) | Nyquist Cutoff | Human Audible Range Coverage | Primary Application Field |
|:---|:---|:---|:---|
| **44,100 Hz (44.1 kHz)** | 22,050 Hz | 100% (Human hearing limit $\approx 20,000$ Hz) | Red Book Audio CD, Consumer Streaming |
| **48,000 Hz (48.0 kHz)** | 24,000 Hz | 100% + Ultrasonic headroom | Cinema, Broadcast TV, YouTube, Opus default |
| **96,000 Hz (96.0 kHz)** | 48,000 Hz | High-Resolution Audio master | Studio mixing, pitch-shifting DSP |
| **192,000 Hz (192.0 kHz)**| 96,000 Hz | Archival / Scientific recording | Specialized mastering |

## 2. Video-to-GIF Quantization & Palette Optimization

Converting MP4 (H.264 / AV1) video streams to GIF files requires reducing 24-bit truecolor space (16.7 million colors) to an 8-bit indexed palette (maximum 256 colors). Without advanced color dithering (Floyd-Steinberg or Hilbert space-filling curve), banding and temporal flicker occur.

Using two-pass palette generation via FFmpeg:
```bash
ffmpeg -i input.mp4 -vf "fps=15,scale=480:-1:flags=lanczos,palettegen=stats_mode=diff" palette.png
ffmpeg -i input.mp4 -i palette.png -lavfi "fps=15,scale=480:-1:flags=lanczos [x]; [x][1:v] paletteuse=dither=bayer:bayer_scale=5" output.gif
```
