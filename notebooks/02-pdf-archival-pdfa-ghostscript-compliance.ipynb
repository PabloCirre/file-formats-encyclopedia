## 1. Conformance Profiles: PDF/A-1 vs PDF/A-2 vs PDF/A-3

| ISO Standard | Level | Base PDF | Key Inclusions | Strict Prohibitions |
|:---|:---:|:---:|:---|:---|
| **ISO 19005-1 (PDF/A-1)** | 1b (Basic)<br>1a (Accessible) | PDF 1.4 | Embedded Type 1 & TrueType fonts, DeviceRGB/CMYK with OutputIntent ICC profile | JavaScript, audio/video streams, encryption, LZW compression |
| **ISO 19005-2 (PDF/A-2)** | 2b / 2u / 2a | PDF 1.7 | JPEG 2000 compression, transparency layers, embedded PDF/A attachments, OpenType fonts | External font references, interactive 3D content, unmanaged color spaces |
| **ISO 19005-3 (PDF/A-3)** | 3b / 3u / 3a | PDF 1.7 | Arbitrary file embedding (XML invoices, CSV datasets, CAD drawings) within document catalog | Missing XMP metadata, non-embedded fonts, unencrypted security locks |

The **`u`** conformance level (e.g., PDF/A-2u) mandates Unicode mappings (`ToUnicode` CMap tables) for every embedded font, ensuring deterministic text searchability and screen-reader accessibility.


```python
# Ghostscript command line recipe for deterministic PDF/A-2b generation
ghostscript_recipe = [
    "gs",
    "-dPDFA=2",
    "-dBATCH",
    "-dNOPAUSE",
    "-dNOOUTERSAVE",
    "-sProcessColorModel=DeviceCMYK",
    "-sColorConversionStrategy=UseDeviceIndependentColor",
    "-sDEVICE=pdfwrite",
    "-dPDFACompatibilityPolicy=1",
    "-dCompatibilityLevel=1.7",
    "-sOutputFile=archived_document_pdfa2b.pdf",
    "PDFA_def.ps",
    "input_source.pdf"
]

print("Ghostscript PDF/A Production Pipeline Invocation:")
print(" ".join(ghostscript_recipe))
```

## 2. VeraPDF Validation Rules & Compliance Verification

A compliant PDF/A file undergoes exhaustive syntactic validation across four primary subsystems:

1. **Font Embedding Verification**: Every font dictionary must contain a valid stream reference (`/FontFile`, `/FontFile2`, or `/FontFile3`). Subsetting (`Tag+FontName`) is permitted provided all referenced glyphs are encoded.
2. **OutputIntent & ICC Profiles**: An `/OutputIntents` array in the document catalog must specify the standardized color gamut (e.g., `sRGB v4` or `ISO Coated v2`).
3. **XMP Extension Schema**: All custom document metadata attributes must be formally declared in the XMP RDF stream with XML namespace URIs.
4. **Flate / JBIG2 Filters**: Lossless compression filters (`/FlateDecode`, `/JBIG2Decode`) must not rely on external dictionaries without embedding.
